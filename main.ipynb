```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio
from scipy.io import wavfile

RATE = 44_100
LA = 440
DO = 523.25

def signal(freq,t):
    return np.sin(2*np.pi*freq*t)

def display_signal(freq):
    t=np.linspace(0,1,RATE)
    son = signal(freq,t)
    plt.plot(t,son)
    plt.show()
    display(Audio(son,rate=RATE))


def sine(freq,duration=1,amplitude=1):
    t=np.linspace(0,duration,duration*RATE)
    son=amplitude*np.sin(2*np.pi*freq*t)
    display(Audio(son,rate=RATE))

#sine(DO,10,5)

def sine_linear(freq1,freq2,duration):
    t=np.linspace(0,duration,duration*RATE)
    freq=freq1+((freq2-freq1)/duration)*t
    son=np.sin(2*np.pi*np.multiply(freq,t))
    display(Audio(son,rate=RATE))

sine_linear(LA,DO,5)








```
