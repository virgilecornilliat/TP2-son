```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio
from scipy.io import wavfile

RATE = 44_100
LA = 440
DO = 523.25

def signal(freq,t):
    return np.sin(2*np.pi*freq*t)

def display_signal(freq):
    t=np.linspace(0,1,RATE)
    son = signal(freq,t)
    plt.plot(t,son)
    plt.show()
    display(Audio(son,rate=RATE))


def sine(freq,duration=1,amplitude=1):
    t=np.linspace(0,duration,duration*RATE)
    son=amplitude*np.sin(2*np.pi*freq*t)
    display(Audio(son,rate=RATE))

#sine(DO,10,5)

def sine_linear(freq1,freq2,duration):
    t=np.linspace(0,duration,duration*RATE)
    freq=freq1+((freq2-freq1)/duration)*t
    son=np.sin(2*np.pi*np.multiply(freq,t))
    display(Audio(son,rate=RATE))

#sine_linear(LA,DO,5)

## 3)
#Pour moduler l'amplitude de la sinusoïde, on créer un tableau d'amplitudes qui suivent une fonction du temps

def crescendo_sine(freq, duration, increase=True):
    t=np.linspace(0,duration,duration*RATE)
    if increase :
        amplitude=np.linspace(0,1,duration*RATE)
    else :
        amplitude=np.linspace(1,0,duration*RATE)
    son=amplitude*np.sin(2*np.pi*np.multiply(freq,t))
    display(Audio(son,rate=RATE))

#crescendo_sine(LA,3,True)
#crescendo_sine(LA,3,False)

def crescendo_sine_2(signal,duration,increase=True):
    t=np.linspace(0,duration,duration*RATE)
    if increase :
        amplitude=np.linspace(0,1,duration*RATE)
    else :
        amplitude=np.linspace(1,0,duration*RATE)
    son=amplitude*signal
    display(Audio(son,rate=RATE))

    
## 4)

#On crée une fonction plus générale ;)

def concatenation(freq1,freq2,duration1,duration2):
    t1=np.linspace(0,duration1,duration1*RATE)
    t2=np.linspace(duration1,duration1+duration2,duration2*RATE)
    son1=np.sin(2*np.pi*freq1*t1)
    son2=np.sin(2*np.pi*freq2*t2)
    son=np.concatenate((son1,son2),axis=0)
    display(Audio(son,rate=RATE))

#concatenation(LA,DO,1,2)

## 5)

def float_to_int16(as_float):
    return np.int64(as_float*32767)

## 6)

ratios=np.array([2**(i/12) for i in range(13)])
def freq_from_name(name):
    gamme={'do':0,'do#':1,'ré':2,'ré#':3,'mi':4,'fa':5,'fa#':6,'sol':7,'sol#':8,'la':9,'la#':10,'si':11}
    return ratios[gamme[name]]*DO/2

#print(np.isclose(freq_from_name('la'),440, 0.1))

## 7)

def superpose(s1,s2,s3):
    assert len(s1)==len(s2) and len(s1)==len(s3)
    return s1+s2+s3

## 8)
def sine_linear_return(freq1,freq2,duration):
    t=np.linspace(0,duration,duration*RATE)
    freq=freq1+((freq2-freq1)/duration)*t
    son=np.sin(2*np.pi*np.multiply(freq,t))
    return son

def concatenation_return(freq1,freq2,duration1,duration2):
    t1=np.linspace(0,duration1,duration1*RATE)
    t2=np.linspace(duration1,duration1+duration2,duration2*RATE)
    son1=np.sin(2*np.pi*freq1*t1)
    son2=np.sin(2*np.pi*freq2*t2)
    son=np.concatenate((son1,son2),axis=0)
    return son


la_do=float_to_int16(superpose(concatenation_return(LA,DO,2,1),sine_linear_return(LA,DO,3),concatenation_return(DO,LA,1,2)))
wavfile.write(filename='la_do',rate=RATE,data=la_do)

RATE2,restored=wavfile.read('la_do')
display(Audio(restored,rate=RATE2))

## 9)







```
